```python
#Importing the required packages
import matplotlib.pyplot as plt
import numpy as np

# Loading the different occupations into a list
occupations = ['Nurse', 'Doctor', 'Engineer', 'Lawyer', 'Teacher',
               'Accountant', 'Salesperson', 'Software Engineer',
               'Scientist', 'Sales Representative', 'Manager']

with_disorder = [64, 7, 6, 5, 31, 7, 30, 1, 2, 2, 0]  # No. of people with sleep disorder for each occupation
total_counts = [73, 71, 63, 47, 40, 37, 32, 4, 4, 2, 1]  # Total no. of people for each occupation
without_disorder = [t - w for t, w in zip(total_counts, with_disorder)]  # No. of people without sleep disorder for each occupation

```

```python
# Creating a new column sleep_final with values 0 if No Sleep Disorder and 1 otherwise
sleep_dataframe["sleep_final"]=np.where(sleep_dataframe["Sleep Disorder"]=="No Sleep Disorder",0,1)
```

```python
# Classifying the Blood Pressure into 5 categories
def classify_bp(Systolic, Diastolic):
    if Systolic < 120 and Diastolic < 80:
        return "Normal"
    elif 120 <= Systolic <= 129 and Diastolic < 80:
        return "Elevated"
    elif 130 <= Systolic <= 139 or 80 <= Diastolic <= 89:
        return "High Blood Pressure Stage 1"
    elif Systolic >= 140 or Diastolic >= 90:
        return "High Blood Pressure Stage 2"
    else:
        return "Unknown"

# Apply the classification function to the dataset
sleep_dataframe['BP_Category'] = sleep_dataframe.apply(lambda row: classify_bp(row['Systolic'], row['Diastolic']), axis=1)
```

```python
#Creating a list for encoding
cols_encode = ["Gender","Occupation","BMI Category","BP_Category"]
```

```python
# Performing One hot encoding
pd.get_dummies(sleep_dataframe[cols_encode])
```

```python
# Preprocessing data for machine learning by one-hot encoding categorical features and removing the target variable column
sleep_dataframe_final =pd.concat([sleep_dataframe.drop(columns=cols_encode),pd.get_dummies(sleep_dataframe[cols_encode])],axis=1).drop(columns="Sleep Disorder")
sleep_dataframe_final
```

```python
sleep_dataframe_final.info()
```

```python
#Importing packages for KNN
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
#Implementing KNN Model

# Step 1: Prepare the data
# Separate features and target variable (assuming the target column is named 'target')
X = sleep_dataframe_final.drop(columns=['Person ID', 'sleep_final'])  # Replace 'target_column_name' with the actual target

# Standardize numerical features using StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Extract the target column (binary or multiclass)
y = sleep_dataframe_final['sleep_final']  # Replace with your target column name

# Step 2: Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)


# Step 3: Train and evaluate K-Nearest Neighbors Classifier
clf_knn = KNeighborsClassifier()
clf_knn.fit(X_train, y_train)
y_pred_knn = clf_knn.predict(X_test)


# Evaluating the model
accuracy = accuracy_score(y_test, y_pred_knn)
precision = precision_score(y_test, y_pred_knn, average='weighted')
recall = recall_score(y_test, y_pred_knn, average='weighted')
f1 = f1_score(y_test, y_pred_knn, average='weighted')
print(" KNN Metrics:")
print(f"Accuracy-----------: {accuracy * 100:.2f}%")
print(f"Precision: {precision * 100:.2f}%")
print(f"Recall: {recall * 100:.2f}%")
print(f"F1 Score: {f1 * 100:.2f}%")
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_knn))

# Confusion matrix
cm = confusion_matrix(y_test, y_pred_knn)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=np.unique(y_test), yticklabels=np.unique(y_test))
plt.title(f" KNN Confusion Matrix ")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

```

```python
#Implementing Decision Tree

# Step 1: Prepare the data
# Separate features and target variable (assuming the target column is named 'target')
X = sleep_dataframe_final.drop(columns=['Person ID', 'sleep_final'])  # Replace 'target_column_name' with the actual target

# Standardize numerical features using StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Extract the target column (binary or multiclass)
y = sleep_dataframe_final['sleep_final']  # Replace with your target column name

# Step 2: Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)


# Step 3: Train and evaluate Decision Tree Classifier
clf_dtf = DecisionTreeClassifier(random_state=42)
clf_dtf.fit(X_train, y_train)
y_pred_dtf = clf_dtf.predict(X_test)

# Evaluating the model

accuracy = accuracy_score(y_test, y_pred_dtf)
precision = precision_score(y_test, y_pred_dtf, average='weighted')
recall = recall_score(y_test, y_pred_dtf, average='weighted')
f1 = f1_score(y_test, y_pred_dtf, average='weighted')
print("Decision Tree Metrics:")
print(f"Accuracy-----------: {accuracy * 100:.2f}%")
print(f"Precision: {precision * 100:.2f}%")
print(f"Recall: {recall * 100:.2f}%")
print(f"F1 Score: {f1 * 100:.2f}%")
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_dtf))

# Confusion matrix
cm = confusion_matrix(y_test, y_pred_dtf)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', xticklabels=np.unique(y_test), yticklabels=np.unique(y_test))
plt.title(f" Decision Tree Confusion Matrix ")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```
